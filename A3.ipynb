```python
import pandas as pd

employee_data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": [
        "2020-03-15",
        "2017-07-19",
        "2013-06-01",
        "2021-02-10",
        "2010-11-25"
    ],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

employees = pd.DataFrame(employee_data)

employees.to_csv("employees.csv", index=False)

print(employees)
```

```python
print(employees.shape)
```

```python
employees.info()
```

```python
print(employees.describe())
```

```python
print("First 5 rows:")
print(employees.head(5))

print("\nLast 3 rows:")
print(employees.tail(3))
```

```python
average_salary = employees["Salary"].mean()

print("Average Salary:", average_salary)
```

```python
total_bonus = employees["Bonus"].sum()

print("Total Bonus:", total_bonus)
```

```python
youngest_age = employees["Age"].min()

print("Youngest Employee Age:", youngest_age)
```

```python
highest_rating = employees["Rating"].max()

print("Highest Rating:", highest_rating)
```

```python
sorted_employees = employees.sort_values(
    by="Salary",
    ascending=False
)

print(sorted_employees)
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

employees["Performance_Category"] = employees["Rating"].apply(
    performance_category
)

print(employees)
```

```python
print(employees.isnull())

print("\nMissing values in each column:")
print(employees.isnull().sum())
```

```python
employees.rename(
    columns={"Employee_ID": "ID"},
    inplace=True
)

print(employees)
```

```python
experienced_employees = employees[
    employees["Years_of_Experience"] > 5
]

print(experienced_employees)
```

```python
it_employees = employees[
    employees["Department"] == "IT"
]

print(it_employees)
```

```python
employees["Tax"] = employees["Salary"] * 0.10

print(employees)
```

```python
employees.to_csv(
    "modified_employees.csv",
    index=False
)

print("Modified dataset saved successfully.")
```
