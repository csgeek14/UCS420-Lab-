```python
# Q2

scores = (45, 89.5, 76, 45.4, 89, 92, 58, 45)

highest = max(scores)
print("Highest score:", highest)
print("Index:", scores.index(highest))

lowest = min(scores)
print("Lowest score:", lowest)
print("Count:", scores.count(lowest))

reversed_scores = list(reversed(scores))
print("Reversed list:", reversed_scores)

x = float(input("Enter score to search: "))

if x in scores:
    print("Score is present")
    print("First occurrence index:", scores.index(x))
else:
    print("Score is not present")
```

```python
# Q3

import random

numbers = [random.randint(100, 900) for i in range(100)]

print("Random Numbers:")
print(numbers)

odd = []
even = []
prime = []

for num in numbers:
    if num % 2 == 0:
        even.append(num)
    else:
        odd.append(num)

    is_prime = True

    if num < 2:
        is_prime = False
    else:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                is_prime = False
                break

    if is_prime:
        prime.append(num)

print("\nOdd Numbers:")
print(odd)
print("Count of odd numbers:", len(odd))

print("\nEven Numbers:")
print(even)
print("Count of even numbers:", len(even))

print("\nPrime Numbers:")
print(prime)
print("Count of prime numbers:", len(prime))
```

```python
# Q4

A = {34, 56, 78, 90}
B = {78, 45, 90, 23}

union = A.union(B)
print("Union:", union)

intersection = A.intersection(B)
print("Intersection:", intersection)

symmetric_difference = A.symmetric_difference(B)
print("Symmetric Difference:", symmetric_difference)

print("Is A subset of B?", A.issubset(B))
print("Is B superset of A?", B.issuperset(A))

x = int(input("Enter score to remove from A: "))

if x in A:
    A.remove(x)
    print("Updated set A:", A)
else:
    print("Score is not present in set A")
```
