```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

corrected_temperature = temperature + 2
print("Original temperatures:", temperature)
print("After adding 2°C:", corrected_temperature)

fahrenheit = (temperature * 9/5) + 32
print("Temperatures in Fahrenheit:", fahrenheit)

greater_than_32 = temperature[temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

count = np.sum(temperature > 32)
print("Number of readings greater than 32°C:", count)

print("\nExplanation:")
print("Vectorization performs operations on the entire NumPy array at once instead of using explicit loops.")
print("Boolean indexing directly selects values satisfying a condition.")
print("Both approaches make the code shorter and generally faster for numerical data processing.")
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

total_steps = np.sum(steps)
print("Total steps:", total_steps)

mean_steps = np.mean(steps)
print("Mean steps:", mean_steps)

maximum = np.max(steps)
minimum = np.min(steps)

print("Maximum steps:", maximum)
print("Minimum steps:", minimum)

day_totals = np.sum(steps, axis=0)
print("Total steps for each day:", day_totals)

user_totals = np.sum(steps, axis=1)
print("Total steps for each user:", user_totals)

max_flat_index = np.argmax(steps)
max_position = np.unravel_index(max_flat_index, steps.shape)

print("Position of maximum steps:", max_position)
print("User:", max_position[0])
print("Day:", max_position[1])
print("Maximum value:", steps[max_position])
```

```python
import numpy as np

original = np.array([1, 2, 3, 4, 5, 6])

print("Original array:", original)

subset = original[1:5]
print("Subset:", subset)

subset[0] = 999

print("\nAfter changing first element of subset to 999:")
print("Original:", original)
print("Subset:", subset)

copied_array = original[1:5].copy()

copied_array[0] = 500

print("\nAfter modifying copied array:")
print("Original:", original)
print("Copied array:", copied_array)

matrix = np.arange(1, 13).reshape(3, 4)

print("\n3 x 4 Matrix:")
print(matrix)

print("\nFirst row:")
print(matrix[0])

print("\nLast row:")
print(matrix[-1])

print("\nSecond column:")
print(matrix[:, 1])

print("\nRows 1-2 and Columns 2-3:")
print(matrix[1:3, 2:4])

flattened = matrix.flatten()
ravelled = matrix.ravel()

print("\nFlattened array:")
print(flattened)

print("\nRavelled array:")
print(ravelled)

ravelled[0] = 100

print("\nAfter modifying ravelled array:")
print("Ravelled:", ravelled)
print("Original matrix:")
print(matrix)

flattened[1] = 200

print("\nAfter modifying flattened array:")
print("Flattened:", flattened)
print("Original matrix:")
print(matrix)

print("\nMatrix Properties:")
print("Shape:", matrix.shape)
print("Number of dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)

print("\nObservation:")
print("A normal slice and ravel() generally return views, so modifying them can affect the original array.")
print("copy() and flatten() create independent copies, so modifying them does not affect the original array.")
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("X:")
print(X)

print("\nShape of X:", X.shape)
print("Dimensions of X:", X.ndim)

X_transpose = X.T

print("\nTranspose of X:")
print(X_transpose)

print("\nTranspose Explanation:")
print("In X, rows represent users and columns represent features.")
print("After transposing, rows represent features and columns represent users.")

XTX = X.T @ X

print("\nX.T @ X:")
print(XTX)

XTX_inverse = np.linalg.inv(XTX)

print("\nInverse of X.T @ X:")
print(XTX_inverse)

beta = np.linalg.inv(X.T @ X) @ X.T @ y

print("\nOLS Coefficients:")
print(beta)

print("\nCoefficient Interpretation:")
print("Coefficient for Sleep Hours:", beta[0])
print("Coefficient for Activity Level:", beta[1])
print("Coefficient for Stress Level:", beta[2])

print("""
Each coefficient represents the estimated change in assistance score
associated with a one-unit increase in that feature, while the other
features are held constant.
""")

new_user = np.array([5, 40, 7])

predicted_score = new_user @ beta

print("New User:", new_user)
print("Predicted Assistance Score:", predicted_score)
```
